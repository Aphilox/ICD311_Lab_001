```python
import torch
import torch.nn as nn
```

```python
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]
```

```python
torch.manual_seed(7)
Xt = torch.tensor(X, dtype=torch.float32)
yt = torch.tensor(y, dtype=torch.float32).view(-1,1)

net = nn.Sequential(nn.Linear(2, 8), nn.ReLU(), nn.Linear(8, 1))
loss_fn = nn.BCEWithLogitsLoss()
opt = torch.optim.Adam(net.parameters(), lr=0.05)

for epoch in range(1000):
    opt.zero_grad()
    logits = net(Xt)
    loss = loss_fn(logits, yt)
    loss.backward()
    opt.step()

with torch.no_grad():
    probs = torch.sigmoid(net(Xt))
    preds = (probs >= 0.5).int()
    print('Probabilities:', probs.squeeze().numpy().round(3))
    print('Predicted labels:', preds.squeeze().numpy())
```
